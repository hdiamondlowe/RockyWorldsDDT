```python
# code to plot stars
def make_context_fig(sample, x_string, y_string, x_label, y_label, color_string='pl_teq_a0', color_label='Planet Equilibrium Temperature (K)', savefig=False):
    fig, ax  = plt.subplots(1, 1 , figsize=(12, 5),
                        constrained_layout=True, sharey=True,
                        sharex=True)
    
    cmap = mpl.cm.inferno
    #norm = mpl.colors.Normalize(vmin=2500, vmax=4000)
    norm = mpl.colors.BoundaryNorm(np.linspace(2500, 4000, 7), cmap.N, extend='both')
    mapper = mpl.cm.ScalarMappable(norm=norm, cmap=cmap)
    
    for i in range(len(sample)):
        
        ax.plot(sample[x_string][i], 
                 sample[y_string][i], 
                 'o',
                 markersize=10,
                 markeredgecolor='grey',
                 color=mapper.to_rgba(sample[color_string][i]),
                 alpha=1.0,
                 zorder=500,
                 )
        ax.text(sample[x_string][i],
             sample[y_string][i], 
             sample['hostname'][i],
             zorder=2000)

    #plt.colorbar(mapper, orientation='vertical', fraction=0.08, pad=0.02, aspect=12).set_label(label=color_label, size=14)
    fig.subplots_adjust(top=1, wspace=1, hspace=0)
    cbar_ax = fig.add_axes([1, 0.1, 0.02, 0.9])
    colorbar = fig.colorbar(mapper, cax=cbar_ax, label=color_label, 
                        location="right", spacing="proportional")      

    ax.set_xlabel(x_label, fontsize=15)
    ax.set_ylabel(y_label, fontsize=15)
    #plt.title('', fontsize=16)

    ax.tick_params(axis='both', labelsize=12)
    ax.tick_params(axis='both', which='minor')#, labelbottom=False)
    ax.grid(axis='both', which='both', alpha=0.4)

    #plt.tick_params(axis='x', which='minor')
    #ax = plt.gca()
    #ax.xaxis.set_major_formatter(StrMethodFormatter('{x:.0f}'))
    #ax.xaxis.set_minor_formatter(StrMethodFormatter('{x:.1f}'))

    #plt.tight_layout()

    #if savefig: plt.savefig(f'../Figure_TerrestrialSample_{x_string}_{y_string}.png', facecolor='white')
    #plt.show()
```

```python
#plot star mass, radius, Teff of MUSCLES and of HotRocksSurvey stars;
#use this to match each sample star to a MUSCLES star for scaling and ETC purposes
%matplotlib inline
make_context_fig(MUSCLES_star_table, 'st_rad', 'st_mass', 
                 'Stellar Radius (R$_\odot$)', 'Stellar Mass (M$_\odot$)',
                 'st_teff', 'Stellar Effective Temperature (K)')

make_context_fig(MUSCLES_star_table, 'sy_dist', 'st_rad', 
                 'System Distance (pc)', 'Stellar Radius (R$_\odot$)',
                 'st_teff', 'Stellar Effective Temperature (K)')

make_context_fig(MUSCLES_star_table, 'st_teff', 'st_rad', 
                 'Stellar Effective Temperature (K)', 'Stellar Radius (R$_\odot$)',
                 'st_teff', 'Stellar Effective Temperature (K)')
```

### Scale MUSCLES spectra

```python
def scale_spectra(comp_spec_filename, comp, targ):

    d_targ   = targ['sy_dist'] * u.pc
    R_targ   = targ['st_rad']  * u.R_sun
    
    d_comp   = comp['sy_dist'] * u.pc
    R_comp   = comp['st_rad']  * u.R_sun

    #d_comp = 2.97 * u.pc
    #R_comp = 0.17 * u.R_sun
    
    comp_spec = fits.getdata(comp_spec_filename)
    flux_comp = comp_spec['FLUX']
    wave_comp = comp_spec['WAVELENGTH']
    flux_comp_err = comp_spec['ERROR']

    flux_targ = flux_comp * ((d_comp/d_targ)**2).decompose() * ((R_targ/R_comp)**2).decompose()
    #flux_targ_err = flux_comp_err * ((d_comp/d_targ)**2).decompose() * ((R_targ/R_comp)**2).decompose()
    
    return wave_comp, flux_comp, np.hstack(flux_targ)
```

```python
def scale_MUSCLES_spectra(targ_sample, MUSCLES_sample, name_targ, name_comp, plot=False):
    
    print(name_comp + '-->' + name_targ)
    
    targ_ind = np.argwhere(name_targ == targ_sample['hostname'])
    name_targ_nospace = name_targ.replace(' ', '')
    
    comp_ind = np.argwhere(name_comp == MUSCLES_sample['hostname'])
    name_comp_small = name_comp.replace(' ', '').lower()
   
    instrument_grating_pairs = np.array([['hst',   'cos',   'g130m',     'component-spec'], 
                                         ['hst',   'cos',   'g160m',     'component-spec'], 
                                         ['hst',   'cos',   'g230l',     'component-spec'], 
                                         ['hst',   'stis',  'g140m',     'component-spec'], 
                                         ['hst',   'stis',  'g230l',     'component-spec'], 
                                         ['hst',   'stis',  'g430l',     'component-spec'], 
                                         #['multi', 'multi', 'broadband', 'var-res-sed']
                                        ])


    for i in instrument_grating_pairs:
        
        comp_spec_filename = f'../MUSCLES/HST_spectra/hlsp_muscles_{i[0]}_{i[1]}_{name_comp_small}_{i[2]}_v22_{i[3]}.fits'
        
        wave_comp, flux_comp, flux_targ = scale_spectra(comp_spec_filename, MUSCLES_sample[comp_ind], targ_sample[targ_ind])
        
        if plot:
            plt.figure(figsize=(15, 6))
            plt.plot(wave_comp, flux_comp, alpha=0.7, label=name_comp)
            plt.plot(wave_comp, flux_targ, alpha=0.7, label=name_targ)
            plt.title(i[1].upper()+' '+i[2].upper())
            plt.legend(loc='best')
            plt.xlabel('Wavelength ($\AA$)')
            plt.ylabel('Flux Density (erg cm$^{-2}$ s$^{-1}$)')
            plt.tight_layout()
            plt.show()

        ascii.write([wave_comp, flux_targ], 
                    f'../MUSCLES/scaled_spectra/{name_targ_nospace}_{i[1].upper()}_{i[2].upper()}_scaledFrom{name_comp_small.upper()}.dat',
                    names=['#wave','fluxden'],
                    overwrite=True)
```

```python
%matplotlib inline
for star in HotRocks:
    if star['hostname'] == 'GJ 357': continue
    if star['hostname'] == 'L 98-59': continue
    scale_MUSCLES_spectra(HotRocks, MUSCLES_star_table, star['hostname'], star['MUSCLEScomp'])
```
