```python
import numpy as np
from astropy.time import Time
import astropy.units as u
from astropy.io import ascii
```

```python
def get_observation_type(targ, obs_start, obs_end, verbose=False):
    '''
    This fough code tries to figure out, given some target information and an observation start and end time,
    what exoplanet event is being observed: transit, eclipse, phase curve, or nothing.

    This code may be buggy and has not been well-tested.

    WARNING: This only makes any sense if you already know what host star is being observing for the observation start/end, 
    and then test the planets in that system. Otherwise, you can put in any time range, and for a large enough sample of planets, 
    you will probably randomly hit on a transit/eclispe/phase curve event. This may be obvious but there you are.
    
    targ --> dictionary: keys/values pulled from NASA Exoplanet Archive
    obs_start --> astropy Time object
    obs_end --> astropy Time object
    verbose --> bool: if True, print more stuff; if Flase, don't
    '''

    planet_name = targ['pl_name']

    print(f"Looking at target {planet_name}")

    P = targ['pl_orbper']*u.day
    T0 = Time((targ['pl_tranmid']*u.day).value, format='jd', scale='utc') 

    # calculate the planetary phase covered by this observation
    n_start = (obs_start - T0).jd*u.day / P
    n_end = (obs_end - T0).jd*u.day / P
    n = int(n_start)

    if verbose:
        print(f"    This observation starts at n={n_start:0.3f} orbits after ephemeris and ends at n={n_end:0.3f} after")

    phase_start = n_start - n
    phase_end = n_end - n

    if verbose:
        print(f"    This observation spans from phase={phase_start:0.3f} to phase={phase_end:0.3f}")

    if phase_end - phase_start > 1: 
        print(f"     This observation probably contains a phase curve of {planet_name}")
    elif phase_start < 0.5 < phase_end:
        print(f"     This observation probably contains a secondary eclipse of {planet_name}")
    elif phase_start < 1.0 < phase_end:
        print(f"     This observation probably contains a transit of {planet_name}")
    else:
        print(f"     This observation probably does not include an event for {planet_name}; or check your parameters")
```
